# Lemmatization

Lemmatization is a text preprocessing technique in Natural Language Processing (NLP) that reduces inflected word forms to their valid dictionary base form, known as a lemma.
Unlike simpler heuristic approaches, lemmatization relies on vocabulary, morphological analysis, and context (such as the word's Part-of-Speech or POS tag) to ensure the resulting base word is a real, meaningful entry in a dictionary.

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "jumps"]

for word in words:
    print(f"{word} --> {lemmatizer.lemmatize(word, pos='v')}") # Specify 'v' for verb to get the correct lemma
```

```python
words_ = ["run", "runner", "running", "ran", "easily", "fairly", "sportingly", "understandable", "playable", "jumps", "jumping", "jumped"]


for word in words_:
    print(f"{word} --> {lemmatizer.lemmatize(word)}") # Default pos is 'n' (noun)
```

```python
print("Noun POS:\n")
for word in words_:
    print(f"{word} --> {lemmatizer.lemmatize(word, pos='n')}") # Specify 'n' for noun to get the correct lemma

print("\nVerb POS:")
for word in words_:
    print(f"{word} --> {lemmatizer.lemmatize(word, pos='v')}") # Specify 'v' for verb to get the correct lemma

print("\nAdjective POS:")
for word in words_:
    print(f"{word} --> {lemmatizer.lemmatize(word, pos='a')}") # Specify 'a' for adjective to get the correct lemma

print("\nAdverb POS:")
for word in words_:
    print(f"{word} --> {lemmatizer.lemmatize(word, pos='r')}") # Specify 'r' for adverb to get the correct lemma
```
