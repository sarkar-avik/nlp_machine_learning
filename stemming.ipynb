# Stemming
Stemming is a text preprocessing technique in Natural Language Processing (NLP) that reduces inflected or derived words down to their root or base form by chopping off prefixes and suffixes.

### PorterStemmer

```python
from nltk.stem import PorterStemmer
from nltk.stem.lancaster import LancasterStemmer
from nltk.stem.snowball import SnowballStemmer

porter_stemmer = PorterStemmer()
lancaster_stemmer = LancasterStemmer()
snowball_stemmer = SnowballStemmer("english")
```

```python
words = ["run", "runner", "running", "eating", "eats", "ate", "ran", "easily", "fairly"]

for word in words:
    print("Porter Stemmer: \n ")
    print(f"{word} --> {porter_stemmer.stem(word)}")


for word in words:
    print("Lancaster Stemmer: \n ")
    print(f"{word} --> {lancaster_stemmer.stem(word)}")


for word in words:
    print("Snowball Stemmer: \n ")
    print(f"{word} --> {snowball_stemmer.stem(word)}")

    
```

```python
from nltk.stem.regexp import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)

reg_words = ["running", "jumps", "understandable", "playable"]

for word in reg_words:
    print(f"{word} --> {reg_stemmer.stem(word)}")
```

### SnowballStemmer

```python
from nltk.stem.snowball import SnowballStemmer

snow_stemmer = SnowballStemmer("english")

reg_words = ["running", "jumps", "understandable", "playable", "fairly", "sportingly"]

print("Snowball Stemmer: \n ")
for word in reg_words:
    print(f"{word} --> {snow_stemmer.stem(word)}")

print("\n----------------\n")
print("Porter Stemmer: \n ")
for word in reg_words:
    print(f"{word} --> {porter_stemmer.stem(word)}")
```
