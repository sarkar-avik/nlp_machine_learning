# Spam Ham using TF-IDF

#### Read Data

```python
import pandas as pd

messages = pd.read_csv('spam_data/SMSSpamCollection.txt', sep='\t', names=['label', 'message'])
```

```python
messages.head()
```

### Data Cleaning and Preprocessing

- Lower Case
- Stopword Removal
- Lemmatization
- Removal of symbols

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re

lemmatizer = WordNetLemmatizer()
```

```python
corpus = []

for i in range(len(messages)):
    message = messages['message'][i]
    
    # remove special characters and make lowercase and split into list of reviews
    reviews = re.sub(r'[^a-zA-Z]', ' ', message).lower().split()    


    # removing stopwords and stemming the words
    reviews = [lemmatizer.lemmatize(word) for word in reviews if word not in set(stopwords.words('english'))]
    reviews = ' '.join(reviews)
    corpus.append(reviews)
```

```python
for i in range(len(corpus)):
    print(f"Original Message :{messages['message'][i]}\nCorpus: {corpus[i]}")
```

# TFIDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=100)
X = tfidf.fit_transform(corpus).toarray()
```

# NGRAM

```python
tfidf = TfidfVectorizer(max_features=100, ngram_range=(1,2))
X = tfidf.fit_transform(corpus).toarray()
```

```python
tfidf.vocabulary_
```

```python
X
```

### Output Feature

```python
y = pd.get_dummies(messages['label'])
```

```python
y
```

Insight:

- This changes spam/ham into 0 and 1.
- Next we will only take the values of either spam/ham that will serve as our output feature
- True is represented by 1 and False by 0 in the underlying layer


```python
y = y.iloc[:,1] # In this case I have considered spam values
```

```python
y
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB

spam_tfidf_model = MultinomialNB()
```

```python
spam_tfidf_model.fit(X_train,y_train)
```

#### Prediction

```python
y_pred = spam_tfidf_model.predict(X_test)
```

## Performance Metrics

```python
from sklearn.metrics import accuracy_score, classification_report

acc_score = accuracy_score(y_test, y_pred)

report = classification_report(y_test, y_pred)
```
